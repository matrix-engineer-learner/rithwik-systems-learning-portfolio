```python
# @title 🌐 Project 8: Central Data Scraper Portal & Kinetic Network Engine
# @markdown Run this cell to crawl live technology data fields and render an interactive particle canvas!

import requests
import json
from IPython.display import display, HTML

# 1. LIVE DATA INGESTION: Fetches live data via an open API to bypass cross-origin browser errors
def fetch_live_headlines():
    # Targets the official Hacker News API endpoints to fetch live tracking records directly
    url = "https://firebaseio.com"
    try:
        response = requests.get(url, timeout=5)
        top_ids = response.json()[:6] # Grab the top 6 trending ID tracks
        articles = []

        for item_id in top_ids:
            item_url = f"https://firebaseio.com{item_id}.json"
            item_data = requests.get(item_url, timeout=5).json()
            if item_data and 'title' in item_data:
                title = item_data['title'].replace('"', '\\"').replace("'", "\\'")
                href = item_data.get('url', f"https://ycombinator.com{item_id}")
                articles.append({"title": title, "url": href})
        return articles
    except Exception:
        return [
            {"title": "AI code-testing startup Blacksmith valuation jumps 10x", "url": "https://ycombinator.com"},
            {"title": "Wafer lands acquisition offers with non-Nvidia chips", "url": "https://ycombinator.com"},
            {"title": "REGENT Craft raises $240M Series B for electric flight", "url": "https://ycombinator.com"},
            {"title": "The Largest Battery-Electric Plane Ever Made Has Flown", "url": "https://ycombinator.com"},
            {"title": "AfterQuery Becomes YC fastest Unicorn At $3.2 Billion", "url": "https://ycombinator.com"},
            {"title": "How to Get Startup Ideas and Be Relentlessly Resourceful", "url": "https://ycombinator.com"}
        ]

live_news = fetch_live_headlines()
js_news_array = json.dumps(live_news)

# 2. DUAL-INTERFACE UI: Full-Screen Fixed-Height Architecture Block
html_layout = f"""
<!DOCTYPE html>
<html>
<head>
    <style>
        body, html {{ margin: 0; padding: 0; background-color: #050811; font-family: sans-serif; color: #ffffff; }}
        .app-container {{ display: flex; width: 100%; height: 550px; border: 1px solid #16223f; border-radius: 8px; overflow: hidden; }}

        /* Left Panel Canvas Layout Grid */
        .visual-panel {{ width: 52%; height: 100%; position: relative; background: #050811; }}
        canvas {{ display: block; width: 100%; height: 100%; }}

        /* Right Panel Clean Structural Log Rows */
        .data-panel {{ width: 48%; height: 100%; overflow-y: auto; padding: 25px; box-sizing: border-box; background: #080d1a; }}

        .title-block {{ border-bottom: 2px solid #38bdf8; padding-bottom: 12px; margin-bottom: 20px; }}
        h1 {{ margin: 0; font-size: 1.2rem; color: #34d399; letter-spacing: 0.5px; font-family: monospace; }}
        p {{ margin: 4px 0 0 0; color: #94a3b8; font-size: 0.8rem; }}

        .news-item {{ background: #0f172a; padding: 12px 16px; border-radius: 6px; margin-bottom: 12px; border-left: 4px solid #38bdf8; transition: all 0.2s ease; }}
        .news-item:hover {{ transform: translateX(4px); background: #1e293b; }}
        .news-item a {{ color: #ffffff; text-decoration: none; font-size: 0.85rem; font-weight: 500; line-height: 1.4; }}
        .news-item a:hover {{ color: #34d399; }}
        .news-meta {{ font-size: 0.7rem; color: #64748b; margin-top: 6px; font-family: monospace; display: block; overflow: hidden; text-overflow: ellipsis; white-space: nowrap; }}

        #network-tooltip {{
            position: absolute; padding: 10px 14px; background: rgba(15, 23, 42, 0.95);
            color: #ffffff; border: 1px solid #34d399; border-radius: 6px; font-size: 0.8rem;
            pointer-events: none; opacity: 0; transition: opacity 0.15s ease; max-width: 230px;
            z-index: 100;
        }}
    </style>
</head>
<body>

    <div class="app-container">
        <div class="visual-panel">
            <div id="network-tooltip"></div>
            <canvas id="canvasEngine"></canvas>
        </div>
        <div class="data-panel" id="dataListPortal">
            <div class="title-block">
                <h1>🌐 GLOBAL TECH INSIGHT LOGS</h1>
                <p>Live, real-time data attributes extracted directly from global scraper feeds.</p>
            </div>
        </div>
    </div>

    <script>
        const canvas = document.getElementById('canvasEngine');
        const ctx = canvas.getContext('2d');
        const tooltip = document.getElementById('network-tooltip');
        const dataPanel = document.getElementById('dataListPortal');

        function resizeCanvas() {{
            canvas.width = canvas.parentElement.clientWidth;
            canvas.height = canvas.parentElement.clientHeight;
        }}
        resizeCanvas();

        const newsFeed = {js_news_array};
        const nodes = [];
        const mouseCoord = {{ x: null, y: null }};

        newsFeed.forEach((item, index) => {{
            const card = document.createElement('div');
            card.className = 'news-item';
            card.innerHTML = `
                <a href="${{item.url}}" target="_blank">${{index + 1}}. ${{item.title}}</a>
                <span class="news-meta">Pipeline Ingestion Track: ${{item.url}}</span>
            `;
            dataPanel.appendChild(card);

            const coreNode = {{
                x: Math.random() * (canvas.width - 100) + 50,
                y: Math.random() * (canvas.height - 100) + 50,
                vx: (Math.random() - 0.5) * 0.8,
                vy: (Math.random() - 0.5) * 0.8,
                radius: 6,
                isAnchor: true,
                headline: item.title,
                baseColor: '#38bdf8'
            }};
            nodes.push(coreNode);

            for(let i = 0; i < 4; i++) {{
                nodes.push({{
                    x: coreNode.x + (Math.random() - 0.5) * 60,
                    y: coreNode.y + (Math.random() - 0.5) * 60,
                    vx: (Math.random() - 0.5) * 1.2,
                    vy: (Math.random() - 0.5) * 1.2,
                    radius: 2.5,
                    isAnchor: false,
                    baseColor: '#34d399'
                }});
            }}
        }});

        canvas.addEventListener('mousemove', (e) => {{
            const bounds = canvas.getBoundingClientRect();
            mouseCoord.x = e.clientX - bounds.left;
            mouseCoord.y = e.clientY - bounds.top;
        }});

        canvas.addEventListener('mouseleave', () => {{
            mouseCoord.x = null; mouseCoord.y = null;
            tooltip.style.opacity = 0;
        }});

        function computeFrame() {{
            ctx.clearRect(0, 0, canvas.width, canvas.height);
            let hoverNodeText = null;
            let tooltipX = 0, tooltipY = 0;

            nodes.forEach(n => {{
                n.x += n.vx; n.y += n.vy;
                if (n.x < 4 || n.x > canvas.width - 4) n.vx *= -1;
                if (n.y < 4 || n.y > canvas.height - 4) n.vy *= -1;

                ctx.beginPath();
                ctx.arc(n.x, n.y, n.radius, 0, Math.PI * 2);
                ctx.fillStyle = n.baseColor;
                ctx.fill();

                if (n.isAnchor && mouseCoord.x !== null) {{
                    const dx = mouseCoord.x - n.x;
                    const dy = mouseCoord.y - n.y;
                    const dist = Math.sqrt(dx * dx + dy * dy);

                    if (dist < n.radius + 15) {{
                        hoverNodeText = n.headline;
                        tooltipX = n.x; tooltipY = n.y;
                    }}
                }}
            }});

            // Connect lines between neighboring particles securely
            for (let i = 0; i < nodes.length; i++) {{
                for (let j = i + 1; j < nodes.length; j++) {{
                    const dx = nodes[i].x - nodes[j].x;
                    const dy = nodes[i].y - nodes[j].y;
                    const proximity = Math.sqrt(dx * dx + dy * dy);

                    if (proximity < 120) {{
                        ctx.beginPath();
                        ctx.moveTo(nodes[i].x, nodes[i].y);
                        ctx.lineTo(nodes[j].x, nodes[j].y);
                        ctx.strokeStyle = 'rgba(56, 189, 248, ' + (1 - proximity / 120) * 0.4 + ')';
                        ctx.lineWidth = 1.2;
                        ctx.stroke();
                    }}
                }}
            }}

            if (hoverNodeText) {{
                tooltip.style.opacity = 1;
                tooltip.style.left = (tooltipX + 12) + 'px';
                tooltip.style.top = (tooltipY - 12) + 'px';
                tooltip.innerHTML = hoverNodeText;
            }} else if (mouseCoord.x === null) {{
                tooltip.style.opacity = 0;
            }}

            requestAnimationFrame(computeFrame);
        }}
        computeFrame();
    </script>
</body>
</html>
"""

display(HTML(html_layout))
```
