```python
# @title 🏥 Capstone Project: Parallel Hospital Triage & Patient Flow Controller
# @markdown Configure metrics for concurrent patient intake lines, then click Play (▶️) to simulate the grid!

# 1. PARALLEL TRACK MEDICAL INTAKE (Left columns of your diagram)
Line_1_Patient_ID = "PATIENT-ER-091" #@param {type:"string"}
Line_1_Vitals_Severity_Score = 82 #@param {type:"slider", min:0, max:100, step:1}
Line_1_Contagion_Risk = True #@param {type:"boolean"}

Line_2_Patient_ID = "PATIENT-AMB-402" #@param {type:"string"}
Line_2_Vitals_Severity_Score = 40 #@param {type:"slider", min:0, max:100, step:1}
Line_2_Contagion_Risk = False #@param {type:"boolean"}

print(f"📥 [CONCURRENT HOSPITAL INTAKE GRID ACTIVE]")
print(f"🚑 Line 1: {Line_1_Patient_ID} (Vitals Severity: {Line_1_Vitals_Severity_Score}/100 | Contagion: {Line_1_Contagion_Risk})")
print(f"🚑 Line 2: {Line_2_Patient_ID} (Vitals Severity: {Line_2_Vitals_Severity_Score}/100 | Contagion: {Line_2_Contagion_Risk})\n")

# 2. SEQUENTIAL DIAGNOSTIC GATES (The columns of diamonds/boxes in your layout)
print("🧠 Processing Clinical Decision Trees...")

def evaluate_patient_flow(patient_id, severity, contagion):
    # Stage 1: Triage Assessment
    if severity > 70:
        status = "🚨 Red Flag: Critical Condition"
        # Stage 2: Isolation & Resource Allocation
        if contagion:
            action = "☣️ OVERRIDE: Divert immediately to Negative-Pressure Isolation Bunker."
        else:
            action = "🏥 URGENT: Route straight to ICU / Trauma Resuscitation Bay."
    else:
        status = "✅ Stable / Yellow Flag"
        if contagion:
            action = "😷 STEP: Route to General Quarantine Waiting Annex."
        else:
            action = "🍏 STANDARD: Direct to Outpatient Clinic / General Ward."
    return status, action

# Run both parallel patient streams simultaneously through the facility logic
l1_status, l1_action = evaluate_patient_flow(Line_1_Patient_ID, Line_1_Vitals_Severity_Score, Line_1_Contagion_Risk)
l2_status, l2_action = evaluate_patient_flow(Line_2_Patient_ID, Line_2_Vitals_Severity_Score, Line_2_Contagion_Risk)

# 3. CONSOLIDATED EMERGENCY DEPT MONITOR REPORT (Right columns)
print("-" * 85)
print(f"📊 CENTRAL LOGISTICS & PATIENT FLOW COMMAND REPORT:")
print(f"▶️ [INTAKE CHANNEL 1] Status: {l1_status} -> Protocol: {l1_action}")
print(f"▶️ [INTAKE CHANNEL 2] Status: {l2_status} -> Protocol: {l2_action}")
print("-" * 85)
```
